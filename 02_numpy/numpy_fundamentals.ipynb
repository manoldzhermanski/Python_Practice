```python
import numpy as np
```

# Task 1 — Create an Array

Create a NumPy array containing:
10, 20, 30, 40, 50

Print:
- the array
- its number of dimensions
- its shape
- its size
- its data type

```python
numbers = np.array([10, 20, 30, 40, 50])
print(numbers)
print(numbers.ndim)
print(numbers.shape)
print(numbers.size)
print(numbers.dtype)
```

# Task 2 — Create Arrays

Create:
- An array of 10 zeros.
- An array of 5 ones.
- An array containing numbers from 0 to 19.
- An array containing even numbers from 2 to 20.
- An array containing 10 equally spaced numbers between 0 and 1.

```python
print(np.zeros(10))
print(np.ones(5))
print(np.arange(0, 20))
print(np.arange(2, 21, 2))
print(np.linspace(0, 1, 10))
```


# Task 3 — Indexing

Given:

**numbers = np.array([15, 23, 8, 42, 17, 31, 9])**

Find:
- First element
- Last element
- Third element
- Second-to-last element
- Element with value 42

```python
numbers = np.array([15, 23, 8, 42, 17, 31, 9])
print(numbers[0])
print(numbers[-1])
print(numbers[2])
print(numbers[-2])
```
