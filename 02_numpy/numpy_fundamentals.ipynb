# Task 10 — Boolean Logic

Given:     
**ages = np.array([18, 25, 31, 17, 42, 29, 16, 35])**

Find:
- Ages ≥ 18
- Ages between 18 and 30
- Ages below 18
- Ages above 30 or below 18
- How many people are between 18 and 30

```python
ages = np.array([18, 25, 31, 17, 42, 29, 16, 35])

above_18 = ages[ages >= 18]
between_18_and_30 = ages[(ages >= 18) & (ages <= 30)]
below_18 = ages[ages < 18]
below_18_or_above_30 = ages[(ages < 18) | (ages > 30)]

print(above_18)
print(between_18_and_30)
print(below_18)
print(below_18_or_above_30)
print(np.size(between_18_and_30))
```

Rows:

```python
employees = np.array([
    [101, 25, 2400],
    [102, 31, 3200],
    [103, 28, 2800],
    [104, 42, 4500],
    [105, 23, 2100],
    [106, 35, 3700],
    [107, 29, 2600],
    [108, 41, 4100]
])

print(f"Average age: {np.average(employees[:, 1])}")
print(f"Average salary: {np.average(employees[:, 2])}")
print(f"Mean age: {np.mean(employees[:, 2])}")
print(f"75 percentile of salary: {np.quantile(employees[:, 2], 0.75)}")
print(f"Employees earning above the median: {employees[employees[:, 2] > np.median(employees[:, 2]), 0]}")
print(f"Employee with the highest salary: {employees[np.argmax(employees[:, 2]), 0]}")
print(f"Employee with the lowest salary: {employees[np.argmin(employees[:, 2]), 0]}")
print(f"Number of employees older than 30: {np.sum(employees[:, 1] > 30)}")
print(f"Number of employees earning more than 3000: {np.sum(employees[:, 2] > 3000)}")
print(f"Salary standard deviation BEFORE salary increase: {np.std(employees[:, 2])}")
employees[np.where(employees[:, 2] < 3000), 2] = employees[np.where(employees[:, 2] < 3000), 2] * 1.1
employees[np.where(employees[:, 2] >= 3000), 2] = employees[np.where(employees[:, 2] >= 3000), 2] * 1.05

print(f"New average salary: {np.average(employees[:, 2])}")
print(f"Salary standard deviation AFTER salary increase: {np.std(employees[:, 2])}")
```
