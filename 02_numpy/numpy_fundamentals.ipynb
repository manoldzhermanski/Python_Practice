# Task 1 — Create an Array

Create a NumPy array containing:
10, 20, 30, 40, 50

Print:
- the array
- its number of dimensions
- its shape
- its size
- its data type

```python
numbers = np.array([10, 20, 30, 40, 50])
print(numbers)
print(numbers.ndim)
print(numbers.shape)
print(numbers.size)
print(numbers.dtype)
```

# Task 2 — Create Arrays

Create:
- An array of 10 zeros.
- An array of 5 ones.
- An array containing numbers from 0 to 19.
- An array containing even numbers from 2 to 20.
- An array containing 10 equally spaced numbers between 0 and 1.

```python
print(np.zeros(10))
print(np.ones(5))
print(np.arange(0, 20))
print(np.arange(2, 21, 2))
print(np.linspace(0, 1, 10))
```


# Task 3 — Indexing

Given:      
**numbers = np.array([15, 23, 8, 42, 17, 31, 9])**

Find:
- First element
- Last element
- Third element
- Second-to-last element
- Element with value 42

```python
numbers = np.array([15, 23, 8, 42, 17, 31, 9])
print(numbers[0])
print(numbers[-1])
print(numbers[2])
print(numbers[-2])
```

# Task 4 — Slicing

Using the same array:     
**numbers = np.array([15, 23, 8, 42, 17, 31, 9])**

Extract:
- First 3 elements
- Last 3 elements
- Every second element
- The array in reverse
- Elements from index 2 to index 5

```python
numbers = np.array([15, 23, 8, 42, 17, 31, 9])
print(numbers[:3])
print(numbers[-3:])
print(numbers[::2])
print(numbers[::-1])
print(numbers[2:6])
```

# Task 5 — 2D Array

Create:     
10 20 30     
40 50 60     
70 80 90     

Then find:
- 50
- 90
- The first row
- The last row
- The second column
- The first two columns
- The bottom-right 2 × 2 matrix

```python
arr = np.array(
    [
        [10, 20, 30],
        [40, 50, 60],
        [70, 80, 90]
    ])

print(arr[1, 1])
print(arr[2, 2])
print(arr[0,])
print(arr[-1,])
print(arr[:,1])
print(arr[:, :2])
print(arr[1:, 1:])
```

# Task 6 — Salary Calculations

salaries = np.array([1800, 2200, 2500, 3100, 2800, 1950, 4200])

Calculate:
- Total salary
- Average salary
- Minimum salary
- Maximum salary
- Median salary
- Standard deviation

**Do not use Python's sum() or min() etc.**

```python
salaries = np.array([1800, 2200, 2500, 3100, 2800, 1950, 4200])
print(np.sum(salaries))
print(np.sum(salaries) / np.size(salaries))
print(np.min(salaries))
print(np.max(salaries))
print(np.median(salaries))
print(np.std(salaries))
```
