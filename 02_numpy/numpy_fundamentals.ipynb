# Task 8 — Different Raises

Using:     
**salaries = np.array([1800, 2200, 2500, 3100, 2800, 1950, 4200])**

- Employees earning less than 2500 receive a 15% raise.
- Employees earning 2500 or more receive a 7% raise.
- Produce the resulting salary array.

**Do this without a loop.**

```python
salaries = np.array([1800, 2200, 2500, 3100, 2800, 1950, 4200])
result = np.where(
    salaries < 2500,
    salaries * 1.15,
    salaries * 1.07
)

print(result)
```

# Task 9 — Filtering

Using:     
**salaries = np.array([1800, 2200, 2500, 3100, 2800, 1950, 4200])**

Find:
- Employees earning more than 2500
- Employees earning less than 2000
- Employees earning between 2000 and 3000
- Number of employees earning more than 2500


```python
salaries = np.array([1800, 2200, 2500, 3100, 2800, 1950, 4200])
above_2500 = salaries[salaries > 2500]
below_2000 = salaries[salaries < 2000]
between_2000_and_3000 = salaries[(salaries >= 2000) & (salaries <= 3000)]

print(above_2500)
print(below_2000)
print(between_2000_and_3000)
print(np.size(above_2500))
```

# Task 10 — Boolean Logic

Given:     
**ages = np.array([18, 25, 31, 17, 42, 29, 16, 35])**

Find:
- Ages ≥ 18
- Ages between 18 and 30
- Ages below 18
- Ages above 30 or below 18
- How many people are between 18 and 30

```python
ages = np.array([18, 25, 31, 17, 42, 29, 16, 35])

above_18 = ages[ages >= 18]
between_18_and_30 = ages[(ages >= 18) & (ages <= 30)]
below_18 = ages[ages < 18]
below_18_or_above_30 = ages[(ages < 18) | (ages > 30)]

print(above_18)
print(between_18_and_30)
print(below_18)
print(below_18_or_above_30)
print(np.size(between_18_and_30))
```

# Task 12 — Categorize Values

Given:     
**scores = np.array([45, 72, 89, 56, 91, 33, 67, 78])**

Create a new array where:
- < 50 → 0
- 50–69 → 1
- 70–84 → 2
- 85+ → 3

For example:
- 45 → 0
- 72 → 2
- 89 → 3

**Hint: np.where() may be useful.**

```python
scores = np.array([45, 72, 89, 56, 91, 33, 67, 78])
result = np.where(scores < 50, 0,
                  np.where(scores < 70, 1,
                           np.where(scores < 85, 2, 3)))

print(result)
```
