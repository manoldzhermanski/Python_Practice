## Task 1

Given:
```python

    "salary": [2500, 3200, 2800, 2100, 4000],
    "department": ["IT", "HR", "IT", "Marketing", "IT"]
```

Write code that displays:
- The number of rows and columns.
- The column names.
- The DataFrame index.
- The data type of each column.

````python
data = {
    "name": ["Ivan", "Maria", "Peter", "Anna", "Georgi"],
    "age": [25, 30, 28, 22, 35],
    "salary": [2500, 3200, 2800, 2100, 4000],
    "department": ["IT", "HR", "IT", "Marketing", "IT"]
}

df = pd.DataFrame(data)

print(f"Number of rows and columns: {df.shape}")
print(f"The column names: {df.columns}")
print(f"The DataFrame index: {df.index}")
print(f"The data type of each column:\n{df.dtypes}")
print(f"The first 3 lines:\n{df.head(3)}")
print(f"The last 2 lines:\n{df.tail(2)}")
print(df.describe())
````

```python
}
```

Create two variables:
- ages = ...
- employee_data = ...

where:
- ages contains only the age column;
- employee_data contains the name and salary columns.

````python
data = {
    "name": ["Ivan", "Maria", "Peter", "Anna", "Georgi"],
    "age": [25, 30, 28, 22, 35],
    "salary": [2500, 3200, 2800, 2100, 4000],
    "department": ["IT", "HR", "IT", "Marketing", "IT"]
}

df = pd.DataFrame(data)

ages = df["age"]
employee_data = df[["age", "salary"]]

print(ages.dtypes)
print(employee_data.dtypes)
````

## Task 39

Given:
```python
df = pd.DataFrame({
    "name": ["Ivan", "Maria", "Peter", "Anna", "Georgi", "Elena", "Nikolay"],
    "age": [25, 30, 28, 22, 35, 27, 31],
    "salary": [2500, 3200, 2800, 2100, 4000, 2900, 3500],
    "department": [
        "IT", "HR", "IT", "Marketing",
        "IT", "HR", "Marketing"
    ]
})
```

- Create a Boolean column `is_senior` which is True if age >= 30, otherwise False.
- Remove the `is_senior` column.
- Rename:   
name → employee_name   
salary → monthly_salary

````python
df = pd.DataFrame({
    "name": ["Ivan", "Maria", "Peter", "Anna", "Georgi", "Elena", "Nikolay"],
    "age": [25, 30, 28, 22, 35, 27, 31],
    "salary": [2500, 3200, 2800, 2100, 4000, 2900, 3500],
    "department": [
        "IT", "HR", "IT", "Marketing",
        "IT", "HR", "Marketing"
    ]
})

df["it_senior"] = False
df.loc[
    df["age"] >= 30,
    "it_senior"
] = True

df = df.drop("it_senior", axis = 1)

df = df.rename(
    columns={
        "name": "employee_name",
        "salary": "monthly_salary"
    }
)

df
````
